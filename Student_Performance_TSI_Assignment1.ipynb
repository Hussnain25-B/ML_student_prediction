```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
math_data = pd.read_csv("student-mat.csv", sep=";")

math_data.head()
```

```python
por_data = pd.read_csv("student-por.csv", sep=";")

por_data.head()
```

```python
print("Mathematics records:", len(math_data))
print("Portuguese records:", len(por_data))
print("Total records:", len(math_data) + len(por_data))
```

```python
math_data["subject"] = "Math"
por_data["subject"] = "Portuguese"
```

```python
data = pd.concat([math_data, por_data], ignore_index=True)

print("Total records:", len(data))
```

```python
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
```

```python
data.columns
```

```python
data["performance"] = np.where(data["G3"] >= 10, "Pass", "Fail")

data[["G3", "performance"]].head(10)
```

```python
print(data["performance"].value_counts())
```

```python
data["performance"].value_counts().plot(kind="bar")

plt.title("Student Performance")
plt.xlabel("Performance")
plt.ylabel("Number of Students")
plt.show()
```
